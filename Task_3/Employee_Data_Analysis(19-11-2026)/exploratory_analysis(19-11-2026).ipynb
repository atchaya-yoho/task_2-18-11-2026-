```python
import pandas as pd

df=pd.read_csv("clean_data.csv")
print(df.head())
```

```python
print(df[["educ","salary","salbegin","jobtime","prevexp"]].agg(["mean","median","min","max","std"]))
```

```python
# groupby
# average salary by jobcat,gender and minority
print(df.groupby("jobcat")["salary"].agg(["mean","median","min","count"]))
print(df.groupby("gender")["salary"].agg(["mean","median","count"]))
print(df.groupby("minority")["salary"].agg(["mean","count"]))
```

```python
# correlation b/w numeric columns
corr=df[["educ","salary","salbegin","jobtime","prevexp"]].corr(numeric_only=True)
print(corr)
```

```python
# top 3 relationships i found
# 1.salbegin and salary-high positive
# 2.educ and salary-medium positive
# 3.prevexp and salary-low
print("top 1: salbegin and salary")
print("top 2: educ and salary")
print("top 3: prevexp and salary")
```

```python
# 1.which jobcat has highest average salary?
print(df.groupby("jobcat")["salary"].mean().sort_values(ascending=False))

# 2.Do males earn more than females on average?
print(df.groupby("gender")["salary"].mean())

#3.Does more education mean more salary?
print(df.groupby("educ")["salary"].mean())
```

```python
# My answers
# 1.Manager has highest average salary
# 2.Yes,male average is higher than female average
# 3.Yes,when educ increases, salary also increases
```
