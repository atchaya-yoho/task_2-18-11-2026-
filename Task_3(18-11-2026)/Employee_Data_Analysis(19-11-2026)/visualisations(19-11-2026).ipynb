```python
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python

# here i making folder for saving figures
os.makedirs("reports/figures",exist_ok=True)
```

```python
df=pd.read_csv("clean_data.csv")
my_color="lightgreen"
```

```python
# bar chart
avg=df.groupby("jobcat")["salary"].mean()

plt.bar(avg.index,avg.values,color=my_color)
plt.title("Average Salary by jobcat")
plt.xlabel("jobcat")
plt.ylabel("Salary in Dollars")
plt.savefig("reports/figures/bar_salary_jobcat.png")
plt.show()
```

# Insight: Manager salary is highest, clerical is lowest 

```python
# Histogram
plt.hist(df["salary"],bins=20,color=my_color,edgecolor="black")
plt.title("Salary Distribution")
plt.xlabel("Salary in Dollars")
plt.ylabel("Number of Employees")
plt.savefig("reports/figures/hist_salary.png")
plt.show()
```

# Most employees earn 22000 to 33000, very few earn high salary.

```python
# scatter plot
plt.scatter(df["salbegin"],df["salary"],color=my_color)
plt.title("Starting Salary vs Current Salary")
plt.xlabel("Starting Salary in Dollars")
plt.ylabel("Current Salary in Dollars")
plt.savefig("reports/figures/scatter_salbegin_salary.png")
plt.show()
```

# when starting salary is high, current salary is also high.

```python
# Line chart
# sorted by educ
avg_educ=df.groupby("educ")["salary"].mean().sort_index()

plt.plot(avg_educ.index,avg_educ.values,marker="o",color="orange")
plt.title("Average Salary by Education Years")
plt.xlabel("Education in Years")
plt.ylabel("Salary in Dollars")
plt.savefig("reports/figures/line_salary_educ.png")
plt.show()
```

# More education means more salary,especially after 16 years it jumps high.

```python
# Box plot
groups=df.groupby("gender")["salary"].apply(list)
labels=list(groups.index)
values=list(groups.values)

print(labels)
print([len(v) for v in values])

plt.boxplot(values,tick_labels=labels)
plt.title("Salary by Gender")
plt.xlabel("Gender")
plt.ylabel("Salary in Dollars")
plt.savefig("reports/figures/box_salary_gender.png")
plt.show()
```

# Male salary is higher than female salary. female has high salary outliers
